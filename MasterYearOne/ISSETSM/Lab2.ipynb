```python
import numpy as np

# Varianta 9 - sectie de spital
# Durata medie ponderata se calculeaza folosind produsul scalar.
durate_medii = np.array([4, 7, 10, 14], dtype=float)
numar_pacienti = np.array([40, 25, 20, 15], dtype=float)

print("Duratele medii de spitalizare:", durate_medii)
print("Numarul pacientilor pe tip de afectiune:", numar_pacienti)

total_pacienti = numar_pacienti.sum()
durata_ponderata_actuala = np.dot(durate_medii, numar_pacienti) / total_pacienti
print("Durata medie ponderata actuala:", durata_ponderata_actuala)

# Scenariul 'ce ar fi daca': al patrulea tip de afectiune creste cu 20%.
numar_pacienti_scenariu = numar_pacienti.copy()
numar_pacienti_scenariu[3] *= 1.20
durata_ponderata_scenariu = np.dot(durate_medii, numar_pacienti_scenariu) / numar_pacienti_scenariu.sum()
print("Numarul de pacienti in scenariu:", numar_pacienti_scenariu)
print("Durata medie ponderata in scenariu:", durata_ponderata_scenariu)
print("Diferenta fata de situatia actuala:", durata_ponderata_scenariu - durata_ponderata_actuala)
```

```python
import numpy as np

# Varianta 11 - tipografie
# Necunoscutele sunt costurile unitare pentru hartie, cerneala si ora de masina.
A = np.array([
    [2.0, 0.5, 3.0],
    [1.0, 0.3, 1.5],
    [0.5, 0.2, 0.8]
])
costuri_comenzi = np.array([465.0, 235.0, 124.0])

print("Matricea consumurilor (hartie, cerneala, timp):")
print(A)
print("Costurile totale facturate pentru carti, brosuri si afise:", costuri_comenzi)
print()

costuri_unitare = np.linalg.solve(A, costuri_comenzi)
print("Costul unitar al hartiei:", costuri_unitare[0])
print("Costul unitar al cernelii:", costuri_unitare[1])
print("Costul unei ore de masina:", costuri_unitare[2])
print("Verificare A @ x:", A @ costuri_unitare)
```

```python
import numpy as np

# Varianta 23 - sistem energetic regional
# Ordinea sectoarelor este: electricitate, termoficare.
# Coloanele matricei A reprezinta consumul intermediar generat de fiecare sector.
A = np.array([
    [0.10, 0.15],
    [0.20, 0.10]
])
cerere_finala = np.array([100.0, 80.0])
I = np.eye(2)
matricea_leontief = I - A

print("Matricea coeficientilor tehnici A:")
print(A)
print("Cererea finala pentru electricitate si termoficare:", cerere_finala)
print("Matricea (I - A):")
print(matricea_leontief)

productie_totala = np.linalg.solve(matricea_leontief, cerere_finala)
print("Productia totala necesara de electricitate:", productie_totala[0])
print("Productia totala necesara de termoficare:", productie_totala[1])
print("Verificare (I - A) @ x:", matricea_leontief @ productie_totala)
```
